```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import os
```

```python
data_path = "audio_features"

files = [f for f in os.listdir(data_path) if f.endswith(".npz")]

all_annotations = []
all_targets = []
all_class_names = None

for f in files:
    d = dict(np.load(os.path.join(data_path, f), allow_pickle=True))
    
    ann = d["annotations"]              # [T, C, A]
    all_annotations.append(ann)
    
    all_targets.append(d["target_classes"])
    
    if all_class_names is None:
        all_class_names = d["class_names"]
```

```python
agreements = []

for ann in all_annotations:
    # mean over annotators
    mean_ann = np.mean(ann, axis=2)
    
    # variance across annotators (disagreement proxy)
    var_ann = np.var(ann, axis=2)
    
    agreements.append(1 - var_ann)

agreements = np.concatenate(agreements, axis=0)

overall_agreement = np.mean(agreements)
print("Overall agreement:", overall_agreement)
```

```python
class_agreement = np.mean(agreements, axis=0)

plt.figure()
plt.bar(range(len(class_agreement)), class_agreement)
plt.xticks(range(len(class_agreement)), all_class_names, rotation=90)
plt.title("Class-wise Annotator Agreement")
plt.show()
```

```python
binary_labels = []

for ann in all_annotations:
    # aggregate annotators
    agg = np.mean(ann, axis=2)
    
    # binarize
    labels = (agg > 0.5).astype(int)
    
    binary_labels.append(labels)

binary_labels = np.concatenate(binary_labels, axis=0)
```

```python
class_freq = np.sum(binary_labels, axis=0)

plt.figure()
plt.bar(range(len(class_freq)), class_freq)
plt.xticks(range(len(class_freq)), all_class_names, rotation=90)
plt.title("Class Frequency")
plt.show()
```

```python
co_occurrence = binary_labels.T @ binary_labels

plt.figure(figsize=(10,8))
sns.heatmap(co_occurrence, xticklabels=all_class_names,
            yticklabels=all_class_names)
plt.title("Class Co-occurrence")
plt.show()
```

```python
durations = np.sum(binary_labels, axis=0)

plt.figure()
plt.bar(range(len(durations)), durations)
plt.xticks(range(len(durations)), all_class_names, rotation=90)
plt.title("Total Event Duration per Class")
plt.show()
```

```python
devices = []
placements = []

for f in files:
    d = dict(np.load(os.path.join(data_path, f), allow_pickle=True))
    devices.append(d["recording_device"])
    placements.append(d["device_placement"])

pd.Series(devices).value_counts().plot(kind="bar", title="Recording Devices")
plt.show()

pd.Series(placements).value_counts().plot(kind="bar", title="Device Placement")
plt.show()
```

```python
mfcc_all = []

for f in files:
    d = dict(np.load(os.path.join(data_path, f), allow_pickle=True))
    mfcc_all.append(d["mfcc_mean"])

mfcc_all = np.concatenate(mfcc_all, axis=0)

print("Mean:", np.mean(mfcc_all))
print("Std:", np.std(mfcc_all))
print("Min:", np.min(mfcc_all))
print("Max:", np.max(mfcc_all))
```

```python
plt.hist(mfcc_all.flatten(), bins=50)
plt.title("MFCC Value Distribution")
plt.show()
```

```python
feature_stack = []

for f in files:
    d = dict(np.load(os.path.join(data_path, f), allow_pickle=True))
    
    features = np.concatenate([
        d["mfcc_mean"],
        d["zcr_mean"],
        d["energy_mean"],
        d["centroid_mean"]
    ], axis=1)
    
    feature_stack.append(features)

feature_stack = np.concatenate(feature_stack, axis=0)

corr = np.corrcoef(feature_stack.T)

plt.figure(figsize=(8,6))
sns.heatmap(corr)
plt.title("Feature Correlation Matrix")
plt.show()
```
